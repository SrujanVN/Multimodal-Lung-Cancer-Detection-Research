```python
# =============================================================================
# IQ-OTH/NCCD Lung Cancer CT Classification – DenseNet-121
# Classes: Normal / Benign / Malignant  +  UNKNOWN (Unsplash)
# Split: 70% Train / 15% Val / 15% Test
# Early Stopping + full metrics + ROC + confusion matrix
# num_workers=0 (stable on Kaggle)
# =============================================================================
# NOTE: These are CT scan slices (originally DICOM → JPEG/PNG on Kaggle),
#       not plain chest X-rays. Converted to RGB for DenseNet-121.
# =============================================================================

import os
import gc
import random
import warnings
from collections import Counter

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset, Subset
from torch.optim.lr_scheduler import CosineAnnealingLR
from torchvision import transforms, models

from sklearn.metrics import (
    classification_report, confusion_matrix, cohen_kappa_score,
    f1_score, precision_score, recall_score, accuracy_score,
    roc_curve, auc
)
from sklearn.preprocessing import label_binarize
from sklearn.model_selection import train_test_split

import kagglehub

warnings.filterwarnings("ignore")
sns.set_style("whitegrid")
plt.rcParams["figure.dpi"] = 120

# ------------------------- 1. DOWNLOAD DATASETS -------------------------
print("Downloading IQ-OTH/NCCD lung cancer CT dataset...")
lung_path = kagglehub.dataset_download("hamdallak/the-iqothnccd-lung-cancer-dataset")
print("CT dataset path:", lung_path)

print("\nDownloading Unsplash (UNKNOWN) dataset...")
unknown_path = kagglehub.dataset_download("lprdosmil/unsplash-random-images-collection")
print("UNKNOWN path:", unknown_path)

# ------------------------- 2. CONFIG -------------------------
SEED          = 42
TRAIN_RATIO   = 0.70
VAL_RATIO     = 0.15
BATCH_SIZE    = 32
NUM_EPOCHS    = 30
IMAGE_SIZE    = 224          # DenseNet standard input size
LEARNING_RATE = 1e-3
WEIGHT_DECAY  = 1e-2
PATIENCE      = 8
MIN_DELTA     = 0.001
MAX_UNKNOWN   = 500

BEST_MODEL_PATH  = "best_lung_ct_densenet121.pth"
FINAL_MODEL_PATH = "final_lung_ct_densenet121.pth"

KAGGLE_INPUT = "/kaggle/input"
UNSPLASH_DIR = "/kaggle/input/datasets/lprdosmil/unsplash-random-images-collection/unsplash-images-collection"

# Official folder names (keep original typo "Bengin")
TISSUE_CLASSES = [
    "Normal cases",
    "Bengin cases",
    "Malignant cases",
]
DISPLAY_NAMES = {
    "Normal cases":    "Normal",
    "Bengin cases":    "Benign",
    "Malignant cases": "Malignant",
    "UNKNOWN":         "UNKNOWN",
}
UNKNOWN_CLASS = "UNKNOWN"
IMG_EXTS = (".png", ".jpg", ".jpeg", ".tiff", ".tif", ".bmp")

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

# ------------------------- 3. REPRODUCIBILITY -------------------------
def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
set_seed()

# ------------------------- 4. PATH HELPERS -------------------------
def find_tissue_root(base=KAGGLE_INPUT):
    markers = ["normal cases", "bengin cases", "malignant cases",
               "normal", "bengin", "malignant", "benign"]
    for root, dirs, _ in os.walk(base):
        hits = sum(1 for m in markers if any(m in d.lower() for d in dirs))
        if hits >= 2:
            print(f"Auto-detected CT dataset at: {root}")
            return root
    if os.path.isdir(lung_path):
        print(f"Using kagglehub path: {lung_path}")
        return lung_path
    raise FileNotFoundError(f"Could not find class folders under {base}")

def find_unknown_root(configured=UNSPLASH_DIR, base=KAGGLE_INPUT):
    if configured and os.path.isdir(configured):
        print(f"Using UNKNOWN folder: {configured}")
        return configured
    for root, _, files in os.walk(base):
        if "unsplash" in root.lower() and any(f.lower().endswith(IMG_EXTS) for f in files):
            print(f"Auto-detected UNKNOWN images at: {root}")
            return root
    print("WARNING: no Unsplash folder found – continuing without UNKNOWN class")
    return None

# ------------------------- 5. DATASET -------------------------
class LungCTDataset(Dataset):
    def __init__(self, root_dir, unknown_dir=None, max_unknown=None):
        self.samples = []
        self.class_counts = Counter()

        has_unknown = unknown_dir is not None and os.path.isdir(unknown_dir)
        self.classes = list(TISSUE_CLASSES)
        if has_unknown:
            self.classes.append(UNKNOWN_CLASS)
        self.class_to_idx = {c: i for i, c in enumerate(self.classes)}
        self.display_names = [DISPLAY_NAMES.get(c, c) for c in self.classes]

        actual_folders = {}
        for root, dirs, files in os.walk(root_dir):
            for d in dirs:
                full = os.path.join(root, d)
                try:
                    if any(f.lower().endswith(IMG_EXTS)
                           for f in os.listdir(full)
                           if os.path.isfile(os.path.join(full, f))):
                        actual_folders[d] = full
                except PermissionError:
                    continue

        for cls_name in TISSUE_CLASSES:
            matched = None
            key = cls_name.split()[0].lower()
            for folder_name, folder_path in actual_folders.items():
                fl = folder_name.lower()
                if (fl == cls_name.lower()
                        or fl.startswith(key)
                        or key in fl
                        or (key == "bengin" and "benign" in fl)):
                    matched = folder_path
                    break
            if matched is None:
                print(f"WARNING: could not find folder for '{cls_name}'")
                continue
            files = [os.path.join(matched, f) for f in os.listdir(matched)
                     if f.lower().endswith(IMG_EXTS)]
            for p in files:
                self.samples.append((p, self.class_to_idx[cls_name]))
                self.class_counts[cls_name] += 1

        if has_unknown:
            files = [os.path.join(unknown_dir, f) for f in os.listdir(unknown_dir)
                     if f.lower().endswith(IMG_EXTS)]
            if max_unknown is not None:
                random.shuffle(files)
                files = files[:max_unknown]
            for p in files:
                self.samples.append((p, self.class_to_idx[UNKNOWN_CLASS]))
                self.class_counts[UNKNOWN_CLASS] += 1

        if not self.samples:
            raise RuntimeError("No CT images found!")
        print("\nClass distribution (CT slices + UNKNOWN):")
        for c, n in self.class_counts.items():
            print(f"  {DISPLAY_NAMES.get(c, c)}: {n}")
        print(f"Total images: {len(self.samples)}")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, label = self.samples[idx]
        img = Image.open(path).convert("RGB")   # CT grayscale → RGB
        return img, label


class TransformSubset(Dataset):
    def __init__(self, subset, transform):
        self.subset = subset
        self.transform = transform
    def __len__(self):
        return len(self.subset)
    def __getitem__(self, idx):
        img, label = self.subset[idx]
        if self.transform:
            img = self.transform(img)
        return img, label

# ------------------------- 6. TRANSFORMS -------------------------
train_tf = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.15, contrast=0.15),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])
val_tf = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])

# ------------------------- 7. MODEL (DenseNet-121) -------------------------
def get_model(num_classes):
    model = models.densenet121(weights=models.DenseNet121_Weights.IMAGENET1K_V1)
    in_features = model.classifier.in_features
    model.classifier = nn.Linear(in_features, num_classes)
    return model

# ------------------------- 8. EARLY STOPPING -------------------------
class EarlyStopping:
    def __init__(self, patience=PATIENCE, min_delta=MIN_DELTA, path=BEST_MODEL_PATH):
        self.patience = patience
        self.min_delta = min_delta
        self.counter = 0
        self.best_score = None
        self.early_stop = False
        self.path = path

    def __call__(self, val_acc, model):
        if self.best_score is None or val_acc > self.best_score + self.min_delta:
            self.best_score = val_acc
            torch.save(model.state_dict(), self.path)
            print(f"  Best model saved (val_acc = {self.best_score:.4f})")
            self.counter = 0
        else:
            self.counter += 1
            print(f"  EarlyStopping counter: {self.counter}/{self.patience}")
            if self.counter >= self.patience:
                self.early_stop = True

# ------------------------- 9. TRAINING UTILITIES -------------------------
def run_epoch(model, loader, criterion, optimizer=None, train=True):
    model.train() if train else model.eval()
    total_loss, correct, total = 0.0, 0, 0
    all_preds, all_labels, all_probs = [], [], []

    with torch.set_grad_enabled(train):
        for x, y in tqdm(loader, leave=False):
            x, y = x.to(DEVICE), y.to(DEVICE)

            out = model(x)
            loss = criterion(out, y)

            if train:
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

            probs = torch.softmax(out, dim=1)
            total_loss += loss.item() * x.size(0)
            preds = out.argmax(dim=1)
            correct += (preds == y).sum().item()
            total += x.size(0)

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(y.cpu().numpy())
            all_probs.extend(probs.detach().cpu().numpy())

    avg_loss = total_loss / total
    avg_acc  = correct / total
    kappa    = cohen_kappa_score(all_labels, all_preds)
    return avg_loss, avg_acc, kappa, np.array(all_preds), np.array(all_labels), np.array(all_probs)

# ------------------------- 10. PLOTS -------------------------
def plot_training_curves(history):
    epochs = range(1, len(history["train_loss"]) + 1)
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    axes[0].plot(epochs, history["train_loss"], label="Train", color="#1f77b4")
    axes[0].plot(epochs, history["val_loss"],   label="Val",   color="#ff7f0e")
    axes[0].set_title("Loss vs Epochs"); axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Loss")
    axes[0].legend(); axes[0].grid(True, alpha=0.3)

    axes[1].plot(epochs, history["train_kappa"], label="Train", color="#1f77b4")
    axes[1].plot(epochs, history["val_kappa"],   label="Val",   color="#ff7f0e")
    axes[1].set_title("Cohen Kappa vs Epochs"); axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Kappa")
    axes[1].legend(); axes[1].grid(True, alpha=0.3)

    axes[2].plot(epochs, [a * 100 for a in history["val_acc"]], color="#2ca02c")
    axes[2].set_title("Validation Accuracy (%)"); axes[2].set_xlabel("Epoch"); axes[2].set_ylabel("Accuracy %")
    axes[2].grid(True, alpha=0.3)

    plt.suptitle("IQ-OTH/NCCD CT – DenseNet-121 Training Curves", fontsize=14)
    plt.tight_layout()
    plt.show()

def plot_confusion_matrix(cm, class_labels):
    plt.figure(figsize=(8, 7))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_labels, yticklabels=class_labels)
    plt.title("Confusion Matrix – CT Lung Cancer (DenseNet-121)", fontsize=13)
    plt.xlabel("Predicted"); plt.ylabel("True")
    plt.tight_layout()
    plt.show()

def plot_roc_curves(y_true, y_score, class_labels):
    n_classes = len(class_labels)
    y_bin = label_binarize(y_true, classes=list(range(n_classes)))
    if n_classes == 2:
        y_bin = np.hstack([1 - y_bin, y_bin])

    plt.figure(figsize=(9, 7))
    for i, name in enumerate(class_labels):
        if y_bin.shape[1] <= i:
            continue
        fpr, tpr, _ = roc_curve(y_bin[:, i], y_score[:, i])
        roc_auc = auc(fpr, tpr)
        plt.plot(fpr, tpr, lw=2, label=f"{name} (AUC = {roc_auc:.3f})")

    plt.plot([0, 1], [0, 1], "k--", lw=1)
    plt.xlim([0.0, 1.0]); plt.ylim([0.0, 1.05])
    plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
    plt.title("ROC Curves (One-vs-Rest) – CT Lung Cancer (DenseNet-121)")
    plt.legend(loc="lower right"); plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# ------------------------- 11. MAIN -------------------------
def main():
    set_seed()

    tissue_root  = find_tissue_root()
    unknown_root = find_unknown_root()

    full_ds = LungCTDataset(
        root_dir=tissue_root,
        unknown_dir=unknown_root,
        max_unknown=MAX_UNKNOWN
    )
    CLASS_NAMES = full_ds.display_names
    num_classes = len(CLASS_NAMES)
    print(f"\nClasses ({num_classes}): {CLASS_NAMES}")

    labels = [s[1] for s in full_ds.samples]
    idx = np.arange(len(full_ds))
    train_idx, temp_idx = train_test_split(
        idx, test_size=(1 - TRAIN_RATIO), stratify=labels, random_state=SEED
    )
    val_ratio_adj = VAL_RATIO / (VAL_RATIO + (1 - TRAIN_RATIO - VAL_RATIO))
    val_idx, test_idx = train_test_split(
        temp_idx, test_size=(1 - val_ratio_adj),
        stratify=[labels[i] for i in temp_idx], random_state=SEED
    )
    print(f"Split → Train: {len(train_idx)} | Val: {len(val_idx)} | Test: {len(test_idx)}")

    train_ds = TransformSubset(Subset(full_ds, train_idx), train_tf)
    val_ds   = TransformSubset(Subset(full_ds, val_idx),   val_tf)
    test_ds  = TransformSubset(Subset(full_ds, test_idx),  val_tf)

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                              num_workers=0, pin_memory=True)
    val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                              num_workers=0, pin_memory=True)
    test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False,
                              num_workers=0, pin_memory=True)

    model     = get_model(num_classes).to(DEVICE)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    scheduler = CosineAnnealingLR(optimizer, T_max=NUM_EPOCHS)
    early_stopper = EarlyStopping()

    history = {
        "train_loss": [], "val_loss": [],
        "train_acc":  [], "val_acc":  [],
        "train_kappa": [], "val_kappa": [],
    }

    print("\n===== TRAINING (CT slices – DenseNet-121) =====")
    for epoch in range(1, NUM_EPOCHS + 1):
        print(f"\nEpoch {epoch}/{NUM_EPOCHS}")
        tr_loss, tr_acc, tr_kappa, *_ = run_epoch(model, train_loader, criterion, optimizer, train=True)
        va_loss, va_acc, va_kappa, *_ = run_epoch(model, val_loader,   criterion, train=False)
        scheduler.step()

        history["train_loss"].append(tr_loss);  history["val_loss"].append(va_loss)
        history["train_acc"].append(tr_acc);    history["val_acc"].append(va_acc)
        history["train_kappa"].append(tr_kappa); history["val_kappa"].append(va_kappa)

        print(f"  Train → Loss: {tr_loss:.4f} | Acc: {tr_acc*100:.2f}% | Kappa: {tr_kappa:.4f}")
        print(f"  Val   → Loss: {va_loss:.4f} | Acc: {va_acc*100:.2f}% | Kappa: {va_kappa:.4f}")

        early_stopper(va_acc, model)
        if early_stopper.early_stop:
            print(f"Early stopping at epoch {epoch}")
            break

        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    if os.path.isfile(BEST_MODEL_PATH):
        model.load_state_dict(torch.load(BEST_MODEL_PATH, map_location=DEVICE))
        print(f"\nLoaded best model from {BEST_MODEL_PATH}")

    print("\n===== TESTING =====")
    model.eval()
    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for inputs, labels in tqdm(test_loader, desc="Testing"):
            inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
            outputs = model(inputs)
            probs = torch.softmax(outputs, dim=1)
            preds = outputs.argmax(dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())

    all_preds  = np.array(all_preds)
    all_labels = np.array(all_labels)
    all_probs  = np.array(all_probs)

    cm = confusion_matrix(all_labels, all_preds)
    print("\nConfusion Matrix:")
    print(cm)

    test_accuracy   = accuracy_score(all_labels, all_preds)
    kappa           = cohen_kappa_score(all_labels, all_preds)
    f1_macro        = f1_score(all_labels, all_preds, average="macro")
    precision_macro = precision_score(all_labels, all_preds, average="macro")
    recall_macro    = recall_score(all_labels, all_preds, average="macro")

    print(f"\nTest Accuracy : {test_accuracy*100:.2f}%")
    print(f"Cohen Kappa   : {kappa:.4f}")
    print(f"Macro F1      : {f1_macro:.4f}")
    print(f"Macro Precision: {precision_macro:.4f}")
    print(f"Macro Recall  : {recall_macro:.4f}")

    print("\nClassification Report:")
    print(classification_report(all_labels, all_preds, target_names=CLASS_NAMES, digits=4))

    print("Per-class Sensitivity / Specificity:")
    for i, name in enumerate(CLASS_NAMES):
        tp = cm[i, i]
        fn = cm[i, :].sum() - tp
        fp = cm[:, i].sum() - tp
        tn = cm.sum() - (tp + fn + fp)
        sens = tp / (tp + fn) if (tp + fn) > 0 else 0.0
        spec = tn / (tn + fp) if (tn + fp) > 0 else 0.0
        print(f"  {name}: Sensitivity={sens:.4f} | Specificity={spec:.4f}")

    plot_training_curves(history)
    plot_confusion_matrix(cm, CLASS_NAMES)
    plot_roc_curves(all_labels, all_probs, CLASS_NAMES)

    torch.save({
        "model_state_dict": model.state_dict(),
        "class_labels": CLASS_NAMES,
        "num_classes": num_classes,
        "image_size": IMAGE_SIZE,
        "history": history,
        "test_acc": test_accuracy,
        "test_kappa": kappa,
        "modality": "CT",
        "backbone": "DenseNet-121",
        "dataset": "IQ-OTH/NCCD + Unsplash UNKNOWN",
    }, FINAL_MODEL_PATH)

    print(f"\nModels saved:")
    print(f"  • {BEST_MODEL_PATH}")
    print(f"  • {FINAL_MODEL_PATH}")


if __name__ == "__main__":
    main()
```
