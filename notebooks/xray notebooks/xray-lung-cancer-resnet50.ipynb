```python
# =============================================================================
# Lung X-ray Classification – ResNet50
# 9 Portuguese classes + UNKNOWN | 70/15/15 | Early Stopping | Full metrics
# =============================================================================

import os, gc, random, warnings
from collections import Counter
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset, Subset
from torch.optim.lr_scheduler import CosineAnnealingLR
from torchvision import transforms, models

from sklearn.metrics import (
    classification_report, confusion_matrix, cohen_kappa_score,
    f1_score, precision_score, recall_score, accuracy_score,
    roc_curve, auc
)
from sklearn.preprocessing import label_binarize
from sklearn.model_selection import train_test_split
import kagglehub

warnings.filterwarnings("ignore")
sns.set_style("whitegrid")
plt.rcParams["figure.dpi"] = 120

# ------------------------- DOWNLOAD -------------------------
print("Downloading lung X-ray dataset...")
lung_path = kagglehub.dataset_download("fernando2rad/x-ray-lung-diseases-images-9-classes")
print("Lung dataset path:", lung_path)

print("\nDownloading Unsplash (UNKNOWN)...")
unknown_path = kagglehub.dataset_download("lprdosmil/unsplash-random-images-collection")
print("UNKNOWN path:", unknown_path)

# ------------------------- CONFIG -------------------------
SEED = 42
BATCH_SIZE = 32
NUM_EPOCHS = 30
IMAGE_SIZE = 224
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-2
DROPOUT_RATE = 0.5
PATIENCE = 8
MIN_DELTA = 0.001
MAX_UNKNOWN = 500

BEST_MODEL_PATH  = "best_lung_resnet50.pth"
FINAL_MODEL_PATH = "final_lung_resnet50.pth"

KAGGLE_INPUT = "/kaggle/input"
UNSPLASH_DIR = "/kaggle/input/datasets/lprdosmil/unsplash-random-images-collection/unsplash-images-collection"

TISSUE_CLASSES = [
    "00 Anatomia Normal", "01 Processos Inflamatórios Pulmonares",
    "02 Maior Densidade", "03 Menor Densidade",
    "04 Doenças Pulmonares Obstrutivas", "05 Doenças Infecciosas Degenerativas",
    "06 Lesões Encapsuladas", "07 Alterações de Mediastino", "08 Alterações do Tórax"
]
UNKNOWN_CLASS = "UNKNOWN"
IMG_EXTS = (".png", ".jpg", ".jpeg", ".tiff", ".tif", ".bmp")

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

def set_seed(seed=SEED):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
set_seed()

# ------------------------- PATH HELPERS -------------------------
def find_tissue_root(base=KAGGLE_INPUT):
    markers = ["00 Anatomia", "01 Processos", "02 Maior", "03 Menor",
               "04 Doenças", "05 Doenças", "06 Lesões", "07 Alterações", "08 Alterações"]
    for root, dirs, _ in os.walk(base):
        if any(any(m in d for m in markers) for d in dirs):
            print(f"Auto-detected lung dataset at: {root}")
            return root
    raise FileNotFoundError(f"Could not find class folders under {base}")

def find_unknown_root(configured=UNSPLASH_DIR, base=KAGGLE_INPUT):
    if configured and os.path.isdir(configured):
        print(f"Using UNKNOWN folder: {configured}")
        return configured
    for root, _, files in os.walk(base):
        if "unsplash" in root.lower() and any(f.lower().endswith(IMG_EXTS) for f in files):
            print(f"Auto-detected UNKNOWN at: {root}")
            return root
    print("WARNING: no Unsplash folder found")
    return None

# ------------------------- DATASET -------------------------
class LungXrayDataset(Dataset):
    def __init__(self, root_dir, unknown_dir=None, max_unknown=None):
        self.samples = []
        self.class_counts = Counter()
        has_unknown = unknown_dir is not None and os.path.isdir(unknown_dir)
        self.classes = list(TISSUE_CLASSES)
        if has_unknown: self.classes.append(UNKNOWN_CLASS)
        self.class_to_idx = {c: i for i, c in enumerate(self.classes)}

        actual_folders = [d for d in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir, d))]
        for cls_name in TISSUE_CLASSES:
            matched = None
            prefix = cls_name[:12]
            for f in actual_folders:
                if f.startswith(prefix) or prefix in f:
                    matched = f; break
            if matched is None:
                print(f"WARNING: could not find folder for '{cls_name}'"); continue
            folder = os.path.join(root_dir, matched)
            files = [os.path.join(folder, f) for f in os.listdir(folder) if f.lower().endswith(IMG_EXTS)]
            for p in files:
                self.samples.append((p, self.class_to_idx[cls_name]))
                self.class_counts[cls_name] += 1

        if has_unknown:
            files = [os.path.join(unknown_dir, f) for f in os.listdir(unknown_dir) if f.lower().endswith(IMG_EXTS)]
            if max_unknown: files = files[:max_unknown]
            for p in files:
                self.samples.append((p, self.class_to_idx[UNKNOWN_CLASS]))
                self.class_counts[UNKNOWN_CLASS] += 1

        print("\nClass distribution:")
        for c, n in self.class_counts.items(): print(f"  {c}: {n}")
        print(f"Total images: {len(self.samples)}")

    def __len__(self): return len(self.samples)
    def __getitem__(self, idx):
        path, label = self.samples[idx]
        return Image.open(path).convert("RGB"), label

class TransformSubset(Dataset):
    def __init__(self, subset, transform):
        self.subset = subset; self.transform = transform
    def __len__(self): return len(self.subset)
    def __getitem__(self, idx):
        img, label = self.subset[idx]
        return self.transform(img) if self.transform else img, label

# ------------------------- TRANSFORMS -------------------------
train_tf = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(), transforms.RandomRotation(15),
    transforms.ColorJitter(0.2, 0.2), transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])
val_tf = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)), transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])

# ------------------------- MODEL -------------------------
def get_model(num_classes):
    model = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V2)
    model.fc = nn.Sequential(
        nn.Dropout(DROPOUT_RATE),
        nn.Linear(model.fc.in_features, num_classes)
    )
    return model

# ------------------------- EARLY STOPPING -------------------------
class EarlyStopping:
    def __init__(self, patience=PATIENCE, min_delta=MIN_DELTA, path=BEST_MODEL_PATH):
        self.patience = patience; self.min_delta = min_delta
        self.counter = 0; self.best_score = None; self.early_stop = False; self.path = path
    def __call__(self, val_acc, model):
        if self.best_score is None or val_acc > self.best_score + self.min_delta:
            self.best_score = val_acc
            torch.save(model.state_dict(), self.path)
            print(f"Best model saved! (val_acc = {self.best_score:.4f})")
            self.counter = 0
        else:
            self.counter += 1
            print(f"EarlyStopping counter: {self.counter}/{self.patience}")
            if self.counter >= self.patience:
                self.early_stop = True

# ------------------------- RUN EPOCH -------------------------
def run_epoch(model, loader, criterion, optimizer=None, train=True):
    model.train() if train else model.eval()
    total_loss, correct, total = 0.0, 0, 0
    all_preds, all_labels, all_probs = [], [], []
    with torch.set_grad_enabled(train):
        for x, y in tqdm(loader, leave=False):
            x, y = x.to(DEVICE), y.to(DEVICE)
            out = model(x)
            loss = criterion(out, y)
            if train:
                optimizer.zero_grad(); loss.backward(); optimizer.step()
            probs = torch.softmax(out, dim=1)
            total_loss += loss.item() * x.size(0)
            preds = out.argmax(1)
            correct += (preds == y).sum().item()
            total += x.size(0)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(y.cpu().numpy())
            all_probs.extend(probs.detach().cpu().numpy())
    return (total_loss/total, correct/total, cohen_kappa_score(all_labels, all_preds),
            np.array(all_preds), np.array(all_labels), np.array(all_probs))

# ------------------------- PLOTS -------------------------
def plot_training_curves(history):
    epochs = range(1, len(history["train_loss"])+1)
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    axes[0].plot(epochs, history["train_loss"], label="Train Loss")
    axes[0].plot(epochs, history["val_loss"], label="Val Loss")
    axes[0].set_title("Loss vs Epochs"); axes[0].legend(); axes[0].grid(True, alpha=0.3)
    axes[1].plot(epochs, history["train_kappa"], label="Train Kappa")
    axes[1].plot(epochs, history["val_kappa"], label="Val Kappa")
    axes[1].set_title("Kappa Score vs Epochs"); axes[1].legend(); axes[1].grid(True, alpha=0.3)
    axes[2].plot(epochs, [a*100 for a in history["val_acc"]], label="Val Accuracy (%)")
    axes[2].set_title("Validation Accuracy vs Epochs"); axes[2].legend(); axes[2].grid(True, alpha=0.3)
    plt.tight_layout(); plt.show()

def plot_confusion_matrix(cm, class_labels):
    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=class_labels, yticklabels=class_labels)
    plt.title("Confusion Matrix Heatmap"); plt.xlabel("Predicted"); plt.ylabel("True")
    plt.xticks(rotation=45, ha="right"); plt.tight_layout(); plt.show()

def plot_roc_curves(y_true, y_probs, class_labels):
    n_classes = len(class_labels)
    y_true_bin = label_binarize(y_true, classes=range(n_classes))
    plt.figure(figsize=(12, 9))
    for i, color in zip(range(n_classes), plt.cm.tab10(np.linspace(0,1,n_classes))):
        fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_probs[:, i])
        plt.plot(fpr, tpr, color=color, lw=2, label=f"{class_labels[i]} (AUC={auc(fpr,tpr):.3f})")
    plt.plot([0,1],[0,1],"k--"); plt.xlabel("FPR"); plt.ylabel("TPR")
    plt.title("Multi-class ROC Curves"); plt.legend(loc="lower right", fontsize=8)
    plt.grid(True, alpha=0.3); plt.tight_layout(); plt.show()

# ------------------------- MAIN -------------------------
def main():
    tissue_root = find_tissue_root()
    unknown_root = find_unknown_root()
    full_ds = LungXrayDataset(tissue_root, unknown_root, MAX_UNKNOWN)
    num_classes = len(full_ds.classes)
    CLASS_NAMES = full_ds.classes
    print(f"\n→ Total classes: {num_classes}")

    labels = [l for _, l in full_ds.samples]
    indices = np.arange(len(full_ds))
    train_idx, temp_idx = train_test_split(indices, test_size=0.30, stratify=labels, random_state=SEED)
    val_idx, test_idx = train_test_split(temp_idx, test_size=0.50,
                                         stratify=[labels[i] for i in temp_idx], random_state=SEED)

    print("\n" + "="*55)
    print("DATASET SPLIT (70% / 15% / 15%)")
    print("="*55)
    print(f"Train: {len(train_idx)} | Val: {len(val_idx)} | Test: {len(test_idx)}")
    print("="*55)

    train_ds = TransformSubset(Subset(full_ds, train_idx), train_tf)
    val_ds   = TransformSubset(Subset(full_ds, val_idx),   val_tf)
    test_ds  = TransformSubset(Subset(full_ds, test_idx),  val_tf)

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=0, pin_memory=True)
    val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=True)
    test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=True)

    model = get_model(num_classes).to(DEVICE)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
    scheduler = CosineAnnealingLR(optimizer, T_max=NUM_EPOCHS)
    early_stopping = EarlyStopping()

    history = {"train_loss":[], "val_loss":[], "train_acc":[], "val_acc":[],
               "train_kappa":[], "val_kappa":[]}

    print("\nStarting training with Early Stopping...")
    for epoch in range(1, NUM_EPOCHS+1):
        print(f"\nEpoch {epoch}/{NUM_EPOCHS}\n" + "-"*50)
        tr_loss, tr_acc, tr_kappa, _, _, _ = run_epoch(model, train_loader, criterion, optimizer, True)
        vl_loss, vl_acc, vl_kappa, _, _, _ = run_epoch(model, val_loader, criterion, None, False)
        scheduler.step()

        history["train_loss"].append(tr_loss); history["val_loss"].append(vl_loss)
        history["train_acc"].append(tr_acc);   history["val_acc"].append(vl_acc)
        history["train_kappa"].append(tr_kappa); history["val_kappa"].append(vl_kappa)

        print(f"Train Loss: {tr_loss:.4f} | Train Kappa: {tr_kappa:.4f}")
        print(f"Val   Loss: {vl_loss:.4f} | Val Kappa: {vl_kappa:.4f} | Val Acc: {vl_acc*100:.2f}%")

        early_stopping(vl_acc, model)
        if early_stopping.early_stop:
            print("Early stopping triggered!"); break

    # ============================ TESTING ============================
    print("\n" + "="*70 + "\nTESTING ON HELD-OUT TEST SET\n" + "="*70)
    model.load_state_dict(torch.load(BEST_MODEL_PATH, map_location=DEVICE))
    model.eval()

    all_preds, all_labels, all_probs = [], [], []
    with torch.no_grad():
        for inputs, labels in tqdm(test_loader, desc="Testing"):
            inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
            outputs = model(inputs)
            probs = torch.softmax(outputs, dim=1)
            preds = outputs.argmax(1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())

    all_preds = np.array(all_preds); all_labels = np.array(all_labels); all_probs = np.array(all_probs)

    cm = confusion_matrix(all_labels, all_preds)
    print("\nConfusion Matrix:\n", cm)
    print(f"\nTest Accuracy: {accuracy_score(all_labels, all_preds)*100:.2f}%")
    print(f"Cohen Kappa: {cohen_kappa_score(all_labels, all_preds):.4f}")
    print(f"Macro F1: {f1_score(all_labels, all_preds, average='macro'):.4f} | "
          f"Macro Precision: {precision_score(all_labels, all_preds, average='macro'):.4f} | "
          f"Macro Recall: {recall_score(all_labels, all_preds, average='macro'):.4f}")
    print("\nClassification Report:")
    print(classification_report(all_labels, all_preds, target_names=CLASS_NAMES, digits=4))

    print("Per-class Sensitivity / Specificity:")
    for i, name in enumerate(CLASS_NAMES):
        tp = cm[i,i]; fn = cm[i,:].sum()-tp; fp = cm[:,i].sum()-tp; tn = cm.sum()-(tp+fn+fp)
        sens = tp/(tp+fn) if (tp+fn)>0 else 0
        spec = tn/(tn+fp) if (tn+fp)>0 else 0
        print(f"  {name}: Sensitivity={sens:.4f} | Specificity={spec:.4f}")

    plot_training_curves(history)
    plot_confusion_matrix(cm, CLASS_NAMES)
    plot_roc_curves(all_labels, all_probs, CLASS_NAMES)

    torch.save({
        "model_state_dict": model.state_dict(),
        "class_labels": CLASS_NAMES,
        "num_classes": num_classes,
        "image_size": IMAGE_SIZE,
        "history": history,
    }, FINAL_MODEL_PATH)
    print(f"\nModels saved:\n  • {BEST_MODEL_PATH}\n  • {FINAL_MODEL_PATH}")

if __name__ == "__main__":
    main()
```
